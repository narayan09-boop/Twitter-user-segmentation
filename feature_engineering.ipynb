# Feature Engineering
This notebook processes base features to extract engagement-based features: tweet count, hashtag usage, and retweets. Standard scaling and log transformations are employed to normalize distributions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv(r'C:\Users\nk090\Desktop\usl project\processed_data.csv', encoding='latin1')
```

```python
# Task 1: Extract engagement-based features (hashtag usage from text)
df['hashtag_count'] = df['text'].astype(str).apply(lambda x: x.count('#'))
```

```python
# Group exactly the engagement-based features per the assignment
features = df[['tweet_count', 'retweet_count', 'fav_number', 'hashtag_count']]

# Twitter engagement metrics often have an extreme long-tail (power law) distribution.
# We apply np.log1p (log(1+x)) to reduce extreme skew.
log_features = np.log1p(features)
```

```python
# Apply standardization
scaler = StandardScaler()
scaled_features = scaler.fit_transform(log_features)
eng_df = pd.DataFrame(scaled_features, columns=features.columns)

eng_df.head()
```

```python
eng_df.to_csv(r'C:\Users\nk090\Desktop\usl project\engineered_features.csv', index=False)
print("Saved feature engineered data to engineered_features.csv")
```
