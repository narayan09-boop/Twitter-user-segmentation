# Feature Embedding
Here we establish compact user representations using Principal Component Analysis (PCA).

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
```

```python
df = pd.read_csv(r'C:\Users\nk090\Desktop\usl project\engineered_features.csv', encoding='latin1')
```

```python
# Task 2: Learn compact user representations using PCA
pca = PCA(n_components=2)
pca_features = pca.fit_transform(df)

pca_df = pd.DataFrame(pca_features, columns=["PC1", "PC2"])
pca_df.head()
```

```python
print("Explained variance ratio:", pca.explained_variance_ratio_)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(pca_df["PC1"], pca_df["PC2"], alpha=0.5, s=10, c='blue')
plt.title("PCA Representation of Users")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()
```

```python
pca_df.to_csv(r'C:\Users\nk090\Desktop\usl project\pca_features.csv', index=False)
print("Saved PCA representations to pca_features.csv")
```
